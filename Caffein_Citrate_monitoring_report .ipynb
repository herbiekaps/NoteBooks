```python


def redcap_long_to_wide(
    df,
    id_col="record_id",
    repeating_forms=None,
    form_col="redcap_repeat_instrument",
    instance_col="redcap_repeat_instance",
    form_variables=None
):
    """
    Convert REDCap repeating instruments from long to wide format.

    Parameters
    ----------
    df : pd.DataFrame
        Original REDCap export.

    id_col : str
        Participant identifier.

    repeating_forms : list or None
        Instruments to reshape.
        None automatically includes all repeating instruments.

    form_col : str
        REDCap repeating instrument column.

    instance_col : str
        REDCap repeating instance column.

    form_variables : dict or None
        Optional dictionary mapping instrument names to variables.
        Recommended when retaining completely empty fields.

    Returns
    -------
    pd.DataFrame
        Dataset with one row per participant.
    """

    df = df.copy()

    # Finding repeating instruments and instances
    df[form_col] = (
        df[form_col]
        .astype("string")
        .str.strip()
        .replace("", pd.NA)
    )

    if repeating_forms is None:
        repeating_forms = df[form_col].dropna().unique().tolist()

    repeating_forms = list(repeating_forms)

    # -----------------------------------------
    # 2. Extract baseline observations
    # -----------------------------------------

    metadata = [id_col, form_col, instance_col]

    baseline = df[df[form_col].isna()].copy()

    if baseline[id_col].duplicated().any():
        raise ValueError(
            "Duplicate baseline records detected. "
            "Check for longitudinal events."
        )

    # Exclude unselected repeating instruments
    repeat_data = df[df[form_col].isin(repeating_forms)].copy()

    # -----------------------------------------
    # 3. Identify repeating variables
    # -----------------------------------------

    if form_variables is None:

        form_variables = {}

        for form in repeating_forms:

            subset = repeat_data[
                repeat_data[form_col] == form
            ]

            form_variables[form] = [
                col for col in df.columns
                if col not in metadata
                and subset[col].notna().any()
            ]

    # Drop repeating variables from baseline
    all_repeat_vars = set(
        var
        for variables in form_variables.values()
        for var in variables
    )

    baseline_vars = [
        col for col in df.columns
        if col not in metadata
        and col not in all_repeat_vars
    ]

    baseline_wide = baseline[
        [id_col] + baseline_vars
    ].copy()

    # -----------------------------------------
    # 4. Reshape each repeating instrument
    # -----------------------------------------

    for form in repeating_forms:

        subset = repeat_data[
            repeat_data[form_col] == form
        ].copy()

        if subset.empty:
            continue

        variables = form_variables.get(form, [])

        if not variables:
            continue

        subset[instance_col] = pd.to_numeric(
            subset[instance_col],
            errors="raise"
        ).astype("Int64")

        if subset[instance_col].isna().any():
            raise ValueError(
                f"Missing repeat instance for {form}"
            )

        if subset.duplicated(
            [id_col, instance_col]
        ).any():
            raise ValueError(
                f"Duplicate repeat instances detected for {form}"
            )

        wide = (
            subset
            .set_index([id_col, instance_col])[variables]
            .unstack(instance_col)
        )

        # Example: followup_1_visit_date
        wide.columns = [
            f"{form}_{instance}_{variable}"
            for variable, instance in wide.columns
        ]

        wide = wide.reset_index()

        baseline_wide = baseline_wide.merge(
            wide,
            on=id_col,
            how="outer",
            validate="one_to_one"
        )

    # -----------------------------------------
    # 5. Preserve participants and validate
    # -----------------------------------------

    # Include all participant IDs, including those without baseline rows
    ids = df[[id_col]].drop_duplicates()

    result = ids.merge(
        baseline_wide,
        on=id_col,
        how="left",
        validate="one_to_one"
    )

    if result[id_col].duplicated().any():
        raise ValueError("Duplicate participant IDs remain.")

    return result

##############################################################################################
# Date difference check function
##############################################################################################

def check_date_difference(
    df,
    participant_id_col,
    start_date_col,
    end_date_col,
    lower_threshold,
    upper_threshold,
):
    """
    Flag rows where the date difference between end_date and start_date is outside
    the allowed [lower_threshold, upper_threshold] range.

    Parameters
    ----------
    df : pd.DataFrame
    participant_id_col : str
        Column with participant ID.
    start_date_col : str
        Column containing the start date.
    end_date_col : str
        Column containing the end date.
    lower_threshold : int
        Minimum allowed number of days difference.
    upper_threshold : int
        Maximum allowed number of days difference.

    Returns
    -------
    pd.DataFrame
        Participant ID, start date, end date, and difference in days
        for rows that are out of bounds.

    Example usage:
    ---------------
    out_of_bounds = check_date_difference(
        df_wide,
        participant_id_col="record_id",
        start_date_col="Start_column",
        end_date_col="End_column",
        lower_threshold=0,
        upper_threshold=365,
    )
    out_of_bounds.head()
    """
    required_cols = [participant_id_col, start_date_col, end_date_col]
    missing = [c for c in required_cols if c not in df.columns]
    if missing:
        raise ValueError(f"Missing required columns: {missing}")

    out = df.loc[:, required_cols].copy()

    # Convert to datetime
    out[start_date_col] = pd.to_datetime(out[start_date_col], errors="coerce", format="%d/%m/%Y")
    out[end_date_col] = pd.to_datetime(out[end_date_col], errors="coerce", format="%d/%m/%Y")

    # Calculate day difference (end - start)
    out["days_difference"] = (out[end_date_col] - out[start_date_col]).dt.days

    # Keep only rows outside the allowed range
    out_of_bounds = out[
        (out["days_difference"] < lower_threshold) |
        (out["days_difference"] > upper_threshold)
    ].copy()

    return out_of_bounds.rename(columns={"days_difference": "days_out_of_bounds"})[
        [participant_id_col, start_date_col, end_date_col, "days_out_of_bounds"]
    ]

##############################################################################################
# Outlier detection function
##############################################################################################

def detect_zscore_outliers(df, value_col, id_col, threshold=3.0):
    """
    Identify outlier records in a single continuous variable using z-scores.

    Parameters
    ----------
    df : pd.DataFrame
        DataFrame containing the continuous variable and record ID.
    value_col : str
        Name of the continuous variable column.
    id_col : str
        Name of the record ID column.
    threshold : float, default=3.0
        Absolute z-score threshold for flagging an outlier.

    Returns
    -------
    pd.DataFrame
        A dataframe with one row per outlier record and columns:
        ['record_id', 'value', 'mean', 'std_dev', 'threshold']
    """
    required_cols = [id_col, value_col]
    missing = [c for c in required_cols if c not in df.columns]
    if missing:
        raise ValueError(f"Missing required columns: {missing}")

    s = pd.to_numeric(df[value_col], errors="coerce")

    mean = s.mean()
    std_dev = s.std(ddof=0)

    if pd.isna(mean) or pd.isna(std_dev) or std_dev == 0:
        return pd.DataFrame(columns=["record_id", "value", "mean", "std_dev", "threshold"])

    z_scores = (s - mean) / std_dev
    outlier_mask = s.notna() & (z_scores.abs() > threshold)

    out = df.loc[outlier_mask, [id_col, value_col]].copy()
    out = out.rename(columns={id_col: "record_id", value_col: "value"})
    out["mean"] = mean
    out["std_dev"] = std_dev
    out["threshold"] = threshold

    return out[["record_id", "value", "mean", "std_dev", "threshold"]].reset_index(drop=True)
```
